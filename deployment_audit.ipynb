```python
import subprocess, sys, json
from pathlib import Path
from datetime import datetime
assert Path('finetuning/config.json').is_file()
SONGMAP_ROOT = None  # Contoh: '/Users/danendra/Documents/ADA26/C6/SongMapApp'
report = Path('outputs.local') / ('deployment-audit-' + datetime.now().strftime('%Y%m%d-%H%M%S') + '.json')
command = [sys.executable, '-u', '-m', 'deployment.audit', '--benchmark',
           '--device', 'mps', '--seconds', '30', '--output', str(report)]
if SONGMAP_ROOT:
    command += ['--songmap-root', SONGMAP_ROOT]
result = subprocess.run(command, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
print(result.stdout)
result.check_returncode()
```
